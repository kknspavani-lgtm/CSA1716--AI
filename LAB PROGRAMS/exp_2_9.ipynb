```python

# Student - Teacher - Subject Database

student_teacher = {
    "anu": ["jai", "ravi"],
    "kiran": ["jai"],
    "latha": ["ravi"]
}

student_subject = {
    "anu": "Python Programming",
    "kiran": "Artificial Intelligence",
    "latha": "Computer Networks"
}

# 1. Find teachers of a student
def taught_by(student):
    return student_teacher.get(student, [])

# 2. Find students of a teacher
def students_of(teacher):
    return [
        student
        for student, teachers in student_teacher.items()
        if teacher in teachers
    ]

# 3. Find classmates of a student
def classmates(student):
    teachers = taught_by(student)
    result = set()

    for teacher in teachers:
        result.update(students_of(teacher))

    result.discard(student)
    return sorted(result)

# 4. Find subject name of a student
def subject_name_of(student):
    return student_subject.get(student)

# Sample queries and outputs
print("?- taught_by(anu, T).")
for t in taught_by("anu"):
    print("T =", t, ";")

print("\n?- students_of(jai, S).")
for s in students_of("jai"):
    print("S =", s, ";")

print("\n?- classmates(anu, X).")
for x in classmates("anu"):
    print("X =", x)

print("\n?- subject_name_of(latha, N).")
print("N =", repr(subject_name_of("latha")))
```
