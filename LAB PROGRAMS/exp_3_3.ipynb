```python

# Family Relationships Database

male = ["rajan", "kumar", "arun", "selvam", "vijay"]
female = ["lakshmi", "meena", "sita", "divya"]

parent = [
    ("rajan", "kumar"),
    ("lakshmi", "kumar"),
    ("rajan", "meena"),
    ("lakshmi", "meena"),
    ("kumar", "arun"),
    ("sita", "arun"),
    ("kumar", "divya"),
    ("sita", "divya"),
    ("meena", "vijay"),
    ("selvam", "vijay")
]

def father(f, c):
    return (f, c) in parent and f in male

def mother(m, c):
    return (m, c) in parent and m in female

def grandparent(g, c):
    return any(
        (g, p) in parent and (p, c) in parent
        for p, child in parent if child == c
    )

def sibling(x, y):
    return x != y and any(
        mother(m, x) and mother(m, y)
        for m in female
    )

def uncle(u, n):
    return u in male and any(
        sibling(u, p) and (p, n) in parent
        for p in set(a for a, b in parent)
    )

def aunt(a, n):
    return a in female and any(
        sibling(a, p) and (p, n) in parent
        for p in set(x for x, y in parent)
    )

def ancestor(a, d):
    if (a, d) in parent:
        return True
    return any(
        ancestor(child, d)
        for p, child in parent if p == a
    )

# Display Prolog-style queries

print("?- father(F, arun).")
for f in male:
    if father(f, "arun"):
        print(f"F = {f}.")

print("\n?- mother(M, vijay).")
for m in female:
    if mother(m, "vijay"):
        print(f"M = {m}.")

print("\n?- grandparent(G, divya).")
for g in ["rajan", "lakshmi", "kumar", "sita", "meena", "selvam"]:
    if grandparent(g, "divya"):
        print(f"G = {g} ;")

print("\n?- sibling(arun, S).")
for s in ["kumar", "meena", "arun", "divya", "vijay"]:
    if sibling("arun", s):
        print(f"S = {s}.")

print("\n?- uncle(U, vijay).")
for u in male:
    if uncle(u, "vijay"):
        print(f"U = {u}.")

print("\n?- aunt(A, arun).")
for a in female:
    if aunt(a, "arun"):
        print(f"A = {a}.")

print("\n?- ancestor(rajan, D).")
for d in ["kumar", "meena", "arun", "divya", "vijay"]:
    if ancestor("rajan", d):
        print(f"D = {d} ;")
```
