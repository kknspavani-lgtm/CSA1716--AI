```python

# Planets Database
# planet(name, position, type, moons)

planets = [
    ("mercury", 1, "rocky", 0),
    ("venus", 2, "rocky", 0),
    ("earth", 3, "rocky", 1),
    ("mars", 4, "rocky", 2),
    ("jupiter", 5, "gas_giant", 95),
    ("saturn", 6, "gas_giant", 146),
    ("uranus", 7, "ice_giant", 28),
    ("neptune", 8, "ice_giant", 16)
]

# 1. gas_giant(P)
print("?- gas_giant(P).")
for p in planets:
    if p[2] == "gas_giant":
        print("P =", p[0], ";")

# 2. closer_than(earth, mars)
print("\n?- closer_than(earth, mars).")
earth_pos = next(p[1] for p in planets if p[0] == "earth")
mars_pos = next(p[1] for p in planets if p[0] == "mars")
print("true." if earth_pos < mars_pos else "false.")

# 3. rocky_planets(L)
print("\n?- rocky_planets(L).")
rocky = [p[0] for p in planets if p[2] == "rocky"]
print("L =", rocky)

# 4. planet(P, _, _, 0)
print("\n?- planet(P, _, _, 0).")
for p in planets:
    if p[3] == 0:
        print("P =", p[0], ";")

# 5. planet(saturn, Pos, Type, Moons)
print("\n?- planet(saturn, Pos, Type, Moons).")
for p in planets:
    if p[0] == "saturn":
        print("Pos =", p[1])
        print("Type =", p[2])
        print("Moons =", p[3])
```
