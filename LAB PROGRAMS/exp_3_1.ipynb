```python
def hanoi(n, from_peg, to_peg, via_peg):
    if n == 1:
        print(f'Move disk 1 from {from_peg} to {to_peg}')
        return

    hanoi(n - 1, from_peg, via_peg, to_peg)
    print(f'Move disk {n} from {from_peg} to {to_peg}')
    hanoi(n - 1, via_peg, to_peg, from_peg)

# Run the algorithm with the requested parameters
hanoi(3, 'left', 'right', 'centre')
```
