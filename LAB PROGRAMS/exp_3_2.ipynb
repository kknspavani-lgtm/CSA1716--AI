```python
# Birds Database

birds = ["sparrow", "eagle", "crow", "parrot", "penguin", "ostrich", "kiwi"]

cannot_fly = ["penguin", "ostrich", "kiwi"]

def can_fly(bird):
    return bird in birds and bird not in cannot_fly

def check(bird):
    if can_fly(bird):
        print(f"{bird} can fly")
    elif bird in birds:
        print(f"{bird} cannot fly")
    else:
        print(f"{bird} is not in the knowledge base")

# Query 1
print("?- can_fly(sparrow).")
print(str(can_fly("sparrow")).lower() + ".")

# Query 2
print("\n?- can_fly(penguin).")
print(str(can_fly("penguin")).lower() + ".")

# Query 3
print("\n?- check(ostrich).")
check("ostrich")
print("true.")

# Query 4
print("\n?- check(dog).")
check("dog")
print("true.")

# Query 5
print("\n?- can_fly(B).")
for bird in birds:
    if can_fly(bird):
        print(f"B = {bird} ;")
```
