```python

def pos(p):
    return p in ["atdoor", "atwindow", "middle"]


def move(state):
    monkey, box, banana, has = state
    moves = []

    # Grasp the banana
    if monkey == "middle" and box == "onbox" and has == "hasnot":
        moves.append(("grasp", (monkey, box, banana, "has")))

    # Climb onto the box
    if monkey == box and box != "onbox" and box in ["atdoor", "atwindow", "middle"]:
        moves.append(("climb", (monkey, "onbox", box, has)))

    # Push the box
    if box != "onbox" and monkey == box:
        for p in ["atdoor", "atwindow", "middle"]:
            if p != box:
                moves.append(
                    (f"push({box},{p})", (p, p, banana, has))
                )

    # Walk to another position
    if box != "onbox":
        for p in ["atdoor", "atwindow", "middle"]:
            if p != monkey:
                moves.append(
                    (f"walk({monkey},{p})", (p, box, banana, has))
                )

    return moves


def plan(state, visited=None):
    if visited is None:
        visited = set()

    if state[3] == "has":
        return []

    visited = visited | {state}

    for action, next_state in move(state):
        if next_state in visited:
            continue

        result = plan(next_state, visited)
        if result is not None:
            return [action] + result

    return None


def solve(start):
    return plan(start)


def show(start):
    result = solve(start)
    print(f"?- solve(state({start[0]}, onfloor, {start[2]}, hasnot), Plan).")
    if result is not None:
        print("Plan = [" + ",".join(result) + "].")
    else:
        print("false.")


show(("atdoor", "atwindow", "atwindow", "hasnot"))
show(("middle", "middle", "middle", "hasnot"))
```
