```python
edges = {
    'a': ['b', 'c'],
    'b': ['d', 'e'],
    'c': ['f'],
    'd': [],
    'e': ['g'],
    'f': ['g'],
    'g': []
}

h = {'a': 7, 'b': 5, 'c': 4, 'd': 6, 'e': 2, 'f': 3, 'g': 0}

def best_first(start, goal):
    frontier = [[start]]
    while frontier:
        path = frontier.pop(0)
        node = path[0]
        if node == goal:
            return list(reversed(path))
        children = [[m] + path for m in edges.get(node, []) if m not in path]
        frontier.extend(children)
        frontier.sort(key=lambda p: h[p[0]])
    return None

print("Path =", best_first('a', 'g'))
print("Path =", best_first('a', 'd'))

```
