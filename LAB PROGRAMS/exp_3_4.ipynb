```python

# Patient Diet Recommendation System

patient = {
    "ravi": "diabetes",
    "meena": "hypertension",
    "kumar": "anaemia",
    "divya": "obesity",
    "arun": "diabetes"
}

diet = {
    "diabetes": "whole grains, oats, green vegetables, pulses, nuts",
    "hypertension": "fruits, vegetables, low-fat dairy, banana, low-salt meals",
    "anaemia": "spinach, beetroot, dates, jaggery, pulses, citrus fruits",
    "obesity": "salads, sprouts, fibre-rich foods, plenty of water, fruits"
}

avoid = {
    "diabetes": "sweets, sugary drinks, white rice in excess",
    "hypertension": "pickles, papad, processed food, excess salt",
    "anaemia": "tea or coffee immediately after meals",
    "obesity": "fried food, fast food, sugary beverages"
}

def suggest(p):
    disease = patient.get(p)
    if disease:
        return diet[disease]
    return None

def advice(p):
    disease = patient.get(p)
    if disease:
        return diet[disease], avoid[disease]
    return None

def with_disease(d):
    return [p for p, disease in patient.items() if disease == d]


# Query 1
print("?- suggest(ravi, D).")
print(f"D = '{suggest('ravi')}'.")

# Query 2
print("\n?- advice(meena, E, A).")
eat, avoid_food = advice("meena")
print(f"E = '{eat}',")
print(f"A = '{avoid_food}'.")

# Query 3
print("\n?- with_disease(diabetes, P).")
for p in with_disease("diabetes"):
    print(f"P = {p} ;")

# Query 4
print("\n?- suggest(divya, D).")
print(f"D = '{suggest('divya')}'.")
```
