```python

from datetime import date

# Database: name and date of birth (day, month, year)
people = {
    "arun": date(2004, 3, 15),
    "chitra": date(2004, 11, 5),
    "divya": date(2003, 3, 20),
    "bala": date(2003, 8, 10),
    "elango": date(2003, 12, 25)
}

# 1. dob(chitra, X)
print("?- dob(chitra, X).")
print("X =", people["chitra"].strftime("%-d/%-m/%Y") + ".")

# 2. born_in_year(N, 2004)
print("\n?- born_in_year(N, 2004).")
for name, dob in people.items():
    if dob.year == 2004:
        print("N =", name, ";")

# 3. born_in_month(N, 3)
print("\n?- born_in_month(N, 3).")
for name, dob in people.items():
    if dob.month == 3:
        print("N =", name, ";")

# 4. older(bala, arun)
print("\n?- older(bala, arun).")
if people["bala"] < people["arun"]:
    print("true.")
else:
    print("false.")

# 5. person(Name, date(_, _, 2003))
print("\n?- person(Name, date(_, _, 2003)).")
for name, dob in people.items():
    if dob.year == 2003:
        print("Name =", name, ";")
```
